```python
# Step 2: Identify and handle missing values.

# Check missing values.

print(df.isnull())
```

```python
# Total missing values 

print(df.isnull().sum().sum())
```

```python
#  Check hidden missing vlaues

print(df.isin(["NA"," ","Unknown","Na",".","?"]).sum())
```

```python
#   Step 3: Remove duplicate records and formatting inconsistencies.

# first identify duplicate records

print(df.duplicated().sum())
```

```python
print(df.columns.tolist())
```

```python
# check formatting inconsistencies (extra space )

for col in df.select_dtypes(include='object').columns:
    print(col, (df[col] != df[col].str.strip()).sum())

```

```python
#remove extra space

df['title']=df['title'].str.strip()
```
